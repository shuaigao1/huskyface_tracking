```python
import serial
import pandas as pd
import altair as alt
import time
from collections import defaultdict
import threading
import ipywidgets as widgets
from IPython.display import display, clear_output, FileLink

# 启用 Altair 渲染器
alt.renderers.enable('default')

# 控制变量
running = False
thread = None
data = []
id_count = defaultdict(int)
csv_filename = ""

# 输出控件
output = widgets.Output()
chart_output = widgets.Output()
# 控制按钮
start_button = widgets.Button(description="开始采集", button_style='success')
stop_button = widgets.Button(description="停止采集", button_style='danger')
download_button = widgets.Button(description="📥 下载 CSV 文件", button_style='info')
download_button.layout.display = 'none'

# 主线程函数
def run_tracking():
    global running, data, id_count, csv_filename

    data = []
    id_count = defaultdict(int)
    csv_filename = ""

    try:
        ser = serial.Serial('COM5', 115200, timeout=1)
        time.sleep(2)
        start_time = time.time()

        with output:
            output.clear_output()
            output.append_stdout("🚀 开始采集（直到手动停止）\n")

        while running:
            line = ser.readline().decode('utf-8').strip()
            if ',' in line:
                parts = line.split(',')
                if len(parts) == 3 and all(p.strip().isdigit() for p in parts):
                    face_id = int(parts[0])
                    x = int(parts[1])
                    y = int(parts[2])

                    id_count[face_id] += 1
                    data.append({"id": str(face_id), "x": x, "y": y})

                    df = pd.DataFrame(data)

                    chart = alt.Chart(df).mark_circle(size=80).encode(
                        x=alt.X('x:Q', scale=alt.Scale(domain=[0, 320])),
                        y=alt.Y('y:Q', scale=alt.Scale(domain=[240, 0])),
                        color='id:N',
                        tooltip=['id', 'x', 'y']
                    ).properties(
                        width=400,
                        height=300,
                        title='HuskyLens Face Tracking'
                    ).interactive()

                    with chart_output:
                        chart_output.clear_output(wait=True)
                        display(chart)
                    with output:
                        output.append_stdout(f"识别 ID:{face_id} 位置: ({x}, {y})\n")
                        output.append_stdout("出现次数统计：\n")
                        for fid, count in id_count.items():
                            output.append_stdout(f"ID {fid}: {count} 次\n")

            time.sleep(0.05)

        ser.close()

        with output:
            output.append_stdout("🛑 已手动停止采集。\n")

            # 保存 CSV
            timestamp = time.strftime("%Y%m%d_%H%M%S")
            csv_filename = f"face_data_{timestamp}.csv"
            pd.DataFrame(data).to_csv(csv_filename, index=False)
            output.append_stdout(f"📁 已保存 CSV：{csv_filename}\n")

            # 显示下载按钮
            download_button.layout.display = 'inline-block'

    except Exception as e:
        with output:
            output.append_stdout(f"❌ 错误：{e}\n")

    running = False

# 事件绑定
def start_clicked(b):
    global running, thread
    if not running:
        running = True
        download_button.layout.display = 'none'
        thread = threading.Thread(target=run_tracking)
        thread.start()

def stop_clicked(b):
    global running
    running = False
    with output:
        output.append_stdout("🧨 手动停止采集...\n")

def download_clicked(b):
    if csv_filename:
        with output:
            display(FileLink(csv_filename, result_html_prefix="👉 点击下载："))

start_button.on_click(start_clicked)
stop_button.on_click(stop_clicked)
download_button.on_click(download_clicked)

# 显示界面
display(widgets.HBox([start_button, stop_button, download_button]))
display(chart_output)
display(output)
```

```python
import serial
import pandas as pd
import altair as alt
import time
from collections import defaultdict
import threading
import ipywidgets as widgets
from IPython.display import display, clear_output, FileLink

# 设置 Altair
alt.renderers.enable('default')
alt.data_transformers.disable_max_rows()

# 全局变量
running = False
thread = None
data = []
id_count = defaultdict(int)
csv_filename = ""

# 控件
output = widgets.Output()
chart_output = widgets.Output()
raw_output = widgets.Output()

start_button = widgets.Button(description="开始采集", button_style='success')
stop_button = widgets.Button(description="停止采集", button_style='danger')
download_button = widgets.Button(description="📥 下载 CSV", button_style='info')
download_button.layout.display = 'none'

# 初始空图
with chart_output:
    df_empty = pd.DataFrame({'x':[160], 'y':[120], 'id':['0']})
    chart = alt.Chart(df_empty).mark_circle(size=80).encode(
        x=alt.X('x:Q', scale=alt.Scale(domain=[0,320])),
        y=alt.Y('y:Q', scale=alt.Scale(domain=[240,0])),
        color='id:N'
    ).properties(width=600, height=400, title='等待数据...')
    display(chart)

# 主函数
def run_tracking():
    global running, data, id_count, csv_filename
    
    data.clear()
    id_count.clear()
    csv_filename = ""
    last_update = 0
    update_interval = 0.3  # 图刷新间隔

    try:
        port = 'COM5'  # 修改成你的端口
        ser = serial.Serial(port, 115200, timeout=1)
        time.sleep(2)
        
        with output:
            output.clear_output()
            print(f"✅ 串口 {port} 已连接，开始采集...")

        while running:
            if ser.in_waiting > 0:
                line = ser.readline().decode('utf-8').strip()
                
                with raw_output:
                    raw_output.clear_output(wait=True)
                    print(f"RAW: {line}")

                if ',' in line:
                    parts = line.split(',')
                    if len(parts) == 3 and all(p.strip().isdigit() for p in parts):
                        face_id = int(parts[0])
                        x = int(parts[1])
                        y = int(parts[2])

                        data.append({"id": str(face_id), "x": x, "y": y})
                        id_count[face_id] += 1

                        with output:
                            output.clear_output(wait=True)
                            print(f"识别 ID:{face_id} 位置: ({x},{y})")
                            print("\n📊 统计:")
                            for fid, count in id_count.items():
                                print(f"ID {fid}: {count} 次")

                        # 控制图表刷新频率
                        now = time.time()
                        if now - last_update > update_interval:
                            df = pd.DataFrame(data[-50:])  # 最近50条
                            chart = alt.Chart(df).mark_circle(size=80).encode(
                                x=alt.X('x:Q', scale=alt.Scale(domain=[0,320])),
                                y=alt.Y('y:Q', scale=alt.Scale(domain=[240,0])),
                                color='id:N',
                                tooltip=['id', 'x', 'y']
                            ).properties(width=600, height=400, title='HuskyLens 实时跟踪').interactive()

                            with chart_output:
                                chart_output.clear_output(wait=True)
                                display(chart)

                            last_update = now

            time.sleep(0.01)

        ser.close()

        # 保存 CSV
        if data:
            csv_filename = f"face_data_{time.strftime('%Y%m%d_%H%M%S')}.csv"
            pd.DataFrame(data).to_csv(csv_filename, index=False)
            with output:
                print(f"✅ 采集完成，数据已保存: {csv_filename}")
            download_button.layout.display = 'inline-block'
    
    except Exception as e:
        with output:
            output.clear_output()
            print(f"❌ 错误: {e}")
        running = False

# 绑定按钮
def start_clicked(b):
    global running, thread
    if not running:
        running = True
        download_button.layout.display = 'none'
        thread = threading.Thread(target=run_tracking)
        thread.start()

def stop_clicked(b):
    global running
    running = False
    with output:
        output.append_stdout("🛑 停止采集...\n")

def download_clicked(b):
    if csv_filename:
        with output:
            display(FileLink(csv_filename, result_html_prefix="⬇️ 点击下载："))

start_button.on_click(start_clicked)
stop_button.on_click(stop_clicked)
download_button.on_click(download_clicked)

# 显示界面
display(widgets.HBox([start_button, stop_button, download_button]))
display(chart_output)
display(output)
display(raw_output)
```
