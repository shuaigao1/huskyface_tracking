### Qn2 Given a 1D Array, A, negate all elements between 2 and 16, in a total of just 2 lines.

```python
a = np.array([1,4,6,8,4,11,34,11,2,9])
a
```

```python
b = np.where((a>2) & (a<16), a*-1, a)
b
```

### Qn3 Write a numpy sequence of commands to normalize a given 10*10 matrix containing integers.

```python
A = np.random.randint(1,10,100).reshape(10,10)
A
```

```python
row_sum = A.sum(axis =1)
row_sum[:,np.newaxis] # this code creates a column vector
```

```python
normedA = A / row_sum[:,np.newaxis]
normedA
```

```python
from sklearn.preprocessing import normalize
normedA2 = normalize(A, axis=1, norm ='l1')
```

### Qn4 Given x +2y +z = 2; 2x+6y+z = 7; x+y+4z = 3. Solve for x,y,z.

```python
A = np.array([[1,2,1], [2,6,1], [1,1,4]])
B = np.array([2,7,3])
result = np.linalg.solve(A,B)
result
```

### Qn5 Write a script that subtracts the mean of all values from each value contained in the array. Futher, report only the values that are larger than 0.

```python
C = np.random.randn(25).reshape(5,5)
C
```

```python
C - C.mean()
```

```python
# Boolean Indexing
C[C-C.mean()>0]
```

### Write a script to find the closest value in a given array based on randomly generated positive scalar value

```python
a = np.array([1.6,7,3,4.5,1,1,8,7.5,3,6,5.5])
```

```python
val = np.random.randint(10)
val
```

```python
np.abs(a -val).argmin()
```

```python
a[np.abs(a -val).argmin()]
```

```python
np.abs(a -val)
```
