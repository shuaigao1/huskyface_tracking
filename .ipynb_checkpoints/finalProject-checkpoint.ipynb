```python
import serial
import pandas as pd
import altair as alt
import time
from collections import defaultdict
from IPython.display import display, clear_output

# 打开Arduino的串口
ser = serial.Serial('COM5', 115200, timeout=1)  # 修改成你的串口号！
time.sleep(2)  # 等待串口稳定

# 初始化数据
data = []
id_count = defaultdict(int)

# 设置Altair渲染器为default（Jupyter自动显示）
alt.renderers.enable('default')

try:
    while True:
        # 读一行串口数据
        line = ser.readline().decode('utf-8').strip()
        if ',' in line:
            try:
                parts = line.split(',')
                face_id = int(parts[0])
                x = int(parts[1])
                y = int(parts[2])

                # 更新统计
                id_count[face_id] += 1

                # 加入数据
                data.append({"id": str(face_id), "x": x, "y": y})

                # 实时更新图表
                df = pd.DataFrame(data)

                chart = alt.Chart(df).mark_circle(size=80).encode(
                    x=alt.X('x:Q', scale=alt.Scale(domain=[0, 320])),
                    y=alt.Y('y:Q', scale=alt.Scale(domain=[240, 0])),  # Y轴反转
                    color='id:N',
                    tooltip=['id', 'x', 'y']
                ).properties(
                    width=400,
                    height=300,
                    title='HuskyLens Face Tracking'
                ).interactive()

                # 清除旧图 + 打印新图
                clear_output(wait=True)
                display(chart)

                # 也可以顺便显示ID出现次数
                print("出现次数统计：")
                for fid, count in id_count.items():
                    print(f"ID {fid}: {count}次")

            except Exception as e:
                print("解析错误:", e)

except KeyboardInterrupt:
    print("\n停止采集。")
    ser.close()
```

```python
import serial
import pandas as pd
import altair as alt
import time
from collections import defaultdict
from IPython.display import display, clear_output
import threading
import ipywidgets as widgets

# 设置Altair渲染器为默认（适用于Jupyter）
alt.renderers.enable('default')

# 初始化变量
ser = serial.Serial('COM5', 115200, timeout=1)  # 修改成你的串口号
time.sleep(2)  # 串口初始化等待
data = []
id_count = defaultdict(int)
running = False  # 控制标志
thread = None

# 开始按钮和停止按钮
start_button = widgets.Button(description="开始采集", button_style='success')
stop_button = widgets.Button(description="停止采集", button_style='danger')

def run_tracking():
    global running
    start_time = time.time()
    while running:
        if time.time() - start_time > 30:
            print("\n已运行 30 秒，自动停止采集。")
            break
        try:
            line = ser.readline().decode('utf-8').strip()
            if ',' in line:
                parts = line.split(',')
                face_id = int(parts[0])
                x = int(parts[1])
                y = int(parts[2])

                id_count[face_id] += 1
                data.append({"id": str(face_id), "x": x, "y": y})

                df = pd.DataFrame(data)
                chart = alt.Chart(df).mark_circle(size=60).encode(
                    x=alt.X('x:Q', scale=alt.Scale(domain=[0, 320])),
                    y=alt.Y('y:Q', scale=alt.Scale(domain=[240, 0])),
                    color='id:N',
                    tooltip=['id', 'x', 'y']
                ).properties(
                    width=400,
                    height=300,
                    title='HuskyLens Face Tracking'
                ).interactive()

                clear_output(wait=True)
                display(widgets.HBox([start_button, stop_button]))
                display(chart)

                print("出现次数统计: ")
                for fid, count in id_count.items():
                    print(f"ID {fid}: {count}次")

            time.sleep(0.05)
        except Exception as e:
            print("解析错误：", e)
    running = False
    ser.close()
    print("串口已关闭。")

def start_clicked(b):
    global running, thread
    if not running:
        running = True
        print("已点击开始按钮，启动采集线程")
        thread = threading.Thread(target=run_tracking)
        thread.start()

def stop_clicked(b):
    global running
    running = False
    print("已请求停止采集...")

start_button.on_click(start_clicked)
stop_button.on_click(stop_clicked)

# 显示按钮
display(widgets.HBox([start_button, stop_button]))
```

```python
import serial
import pandas as pd
import altair as alt
import time
from collections import defaultdict
import threading
import ipywidgets as widgets
from IPython.display import display, clear_output

# 设定 Altair 显示
alt.renderers.enable('default')

# 控制变量
running = False
thread = None
data = []
id_count = defaultdict(int)

# 创建输出区域
output = widgets.Output()

# 定义按钮
start_button = widgets.Button(description="开始采集", button_style='success')
stop_button = widgets.Button(description="停止采集", button_style='danger')

# 主运行函数（采集线程用）
def run_tracking():
    global running, data, id_count

    data = []
    id_count = defaultdict(int)

    try:
        ser = serial.Serial('COM5', 115200, timeout=1)  # 修改成你的串口号
        time.sleep(2)

        with output:
            print("🚀 开始采集数据...（自动运行30秒）")

        start_time = time.time()

        while running and (time.time() - start_time <= 30):
            line = ser.readline().decode('utf-8').strip()
            if ',' in line:
                try:
                    parts = line.split(',')
                    if len(parts) >= 3:
                        face_id = int(parts[0])
                        x = int(parts[1])
                        y = int(parts[2])

                        id_count[face_id] += 1
                        data.append({"id": str(face_id), "x": x, "y": y})

                        df = pd.DataFrame(data)
                        chart = alt.Chart(df).mark_circle(size=60).encode(
                            x=alt.X('x:Q', scale=alt.Scale(domain=[0, 320])),
                            y=alt.Y('y:Q', scale=alt.Scale(domain=[240, 0])),
                            color='id:N',
                            tooltip=['id', 'x', 'y']
                        ).properties(
                            width=400,
                            height=300,
                            title='HuskyLens Face Tracking'
                        ).interactive()

                        with output:
                            clear_output(wait=True)
                            display(widgets.HBox([start_button, stop_button]))
                            display(chart)
                            print("🧠 出现次数统计：")
                            for fid, count in id_count.items():
                                print(f"ID {fid}: {count} 次")

                    time.sleep(0.05)

                except Exception as e:
                    with output:
                        print("❌ 解析错误：", e)

        ser.close()
        with output:
            print("🛑 自动停止采集，串口已关闭。")

    except Exception as e:
        with output:
            print("❌ 串口初始化失败：", e)

    running = False  # 确保状态正确关闭

# 开始按钮回调
def start_clicked(b):
    global running, thread
    if not running:
        running = True
        thread = threading.Thread(target=run_tracking)
        thread.start()

# 停止按钮回调
def stop_clicked(b):
    global running
    running = False
    with output:
        print("🧨 已点击手动停止采集。")

# 绑定按钮事件
start_button.on_click(start_clicked)
stop_button.on_click(stop_clicked)

# 显示按钮和输出区域
display(widgets.HBox([start_button, stop_button]))
display(output)
```
